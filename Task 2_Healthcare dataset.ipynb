```python
df.drop_duplicates(inplace=True)
```

```python
df.duplicated().sum()
```

```python
df.columns = (
    df.columns
      .str.strip()
      .str.lower()
      .str.replace(" ", "_")
)
```

```python
df.columns
```

```python
df.info()
```

```python
df['date_of_admission'] = pd.to_datetime(df['date_of_admission'])
```

```python
df['discharge_date'] = pd.to_datetime(df['discharge_date'])
```

```python
df.info()
```

```python
df.fillna("Unknown", inplace=True)
df['doctor'].fillna('Unknown', inplace=True)
df['billing_amount'].fillna(
    df['billing_amount'].median(),
    inplace=True
)
```

```python
df.isnull().sum()
```

```python
Q1 = df['billing_amount'].quantile(0.25)
Q3 = df['billing_amount'].quantile(0.75)

IQR = Q3 - Q1
```

```python
lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR
```

```python
outliers = df[
    (df['billing_amount'] < lower) |
    (df['billing_amount'] > upper)
]
```

```python
outliers.shape
```

```python
df = df[
    (df['billing_amount'] >= lower) &
    (df['billing_amount'] <= upper)
]
```

```python
df.shape
```

```python
df.info()
df.isnull().sum()
df.duplicated().sum()
df.describe()
```

```python
df.to_csv(
    "cleaned_healthcare_dataset.csv",
    index=False
)
```
